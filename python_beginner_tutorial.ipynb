## 6. Listen

```python
zahlen = [1, 2, 3, 4, 5]

print(zahlen[0])
zahlen.append(6)

for zahl in zahlen:
    print(zahl)
```

## 8. Mini-Projekt: Zahlen raten

```python
import random

geheimzahl = random.randint(1, 10)
tipp = int(input("Rate eine Zahl zwischen 1 und 10: "))

if tipp == geheimzahl:
    print("Richtig!")
else:
    print("Falsch! Die Zahl war:", geheimzahl)
```
