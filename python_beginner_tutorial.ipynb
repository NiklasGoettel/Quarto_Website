## 9💻 Implementing a little algorithm 

-> the computer should guess your number between 1 and 100

```python
import random
guess=False
initial=1
upper=100
while guess==False:
    computer=random.randint(initial,upper)
    user=str(input(f"Is your number bigger than {computer}?"))
    if user=="yes":
        initial=computer
     
    elif user=="exact":
        guess=True
    else: 
        upper=computer
```
