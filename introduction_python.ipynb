### 4.4. Dictionaries

Dictionaries are unordered collections of key-value pairs, where each unique key maps to a specific value. They are defined using curly brackets {}.


```python
empty_dict={} # initiating a dict


names=["Tom","Jan"]
ages=[20,30]

for name, age in zip(names,ages):  # a zip function is introduced here. It is useful if you want to iterate over several lists or tuples..
    empty_dict[name]=age # name=key, age=values 
```

```python
empty_dict
```

Dictionaries are clever data structures which have a lot of inbuilt function.
One of the most useful might be as follows:

```python
age_list=empty_dict.values() # getting the values from a dictionary
age_list
```

```python
empty_dict.keys() #getting the keys 
```

Also note that dict() can work with all kind of data (bools, int, float)



```python
temperature=[1,2,4,14,1]

days=["Monday","Tuesday","Saturday"]

weather_data={}

for temp,d in zip(temperature,days):
    weather_data[d]=f"{temp} degrees" # the f"" text {variable}" is a convenient way to combine generic variable values and text and has a lot of practice
weather_data
```

### 6. Generators 

They are sometimes useful if you want to create an iterable object by your own and especially if you deal with big data (it is a way to define how to iterate).
The syntax is quite similiar to a function, yet we need the "yield" statement. 
Generators are lazy i.e., the just execute if demanded.

```python
# Example generator

def squared(x=10): # also note that the default argument of the function is fixed to 10 
    for i in range(x):
        yield i**2
```

```python
generator=squared()
# note that is safes CPU capacity
for i in generator:
    print(i)
```

```python
# the difference in memory usage 

import sys

lst = [i for i in range(1_000_000)]
gen = (i for i in range(1_000_000)) # similiar to list comprehension but with parentheses

print(sys.getsizeof(lst))
print(sys.getsizeof(gen))

```

There is also a specific library for generators

```python
import itertools
```

```python
def first_letter(x):
    return x[0]

names=["Marvin","Jan","Silvia"]


for letter,names in itertools.groupby(names,first_letter): # the letters s
    print(letter)

```

```python


# Example data
companies = [
    "A", "B", "C", "D", "E",
    "F", "G", "H", "I", "J",
    "K", "L"
]

advertising_spend = np.array([
    20, 35, 15, 50, 42,
    28, 60, 75, 45, 30,
    80, 55
])  # in thousand €

revenue = np.array([
    120, 150, 95, 210, 180,
    130, 260, 310, 195, 145,
    340, 230
])  # in thousand €

employees = np.array([
    20, 35, 15, 50, 80,
    25, 120, 150, 70, 40,
    180, 100
])

industries = np.array([
    "IT", "IT", "Retail", "IT",
    "Retail", "Services", "IT",
    "Retail", "Services", "Retail",
    "IT", "Services"
])

# Colors for each industry
colors = {
    "IT": "royalblue",
    "Retail": "orange",
    "Services": "seagreen"
}

# Assign a color to each company
point_colors = [colors[industry] for industry in industries]

# Calculate point sizes based on number of employees
point_sizes = employees * 8

# Create the scatter plot
plt.figure(figsize=(11, 7))

plt.scatter(
    advertising_spend,
    revenue,
    s=point_sizes,
    c=point_colors,
    alpha=0.7,
    edgecolors="black",
    linewidths=1
)

# Add company labels
for i, company in enumerate(companies):
    plt.annotate(
        company,
        (advertising_spend[i], revenue[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

# Create legend for industries
for industry, color in colors.items():
    plt.scatter(
        [],
        [],
        s=150,
        c=color,
        label=industry,
        edgecolors="black"
    )

# Create legend for point sizes
for employee_count in [25, 75, 150]:
    plt.scatter(
        [],
        [],
        s=employee_count * 8,
        c="gray",
        alpha=0.5,
        edgecolors="black",
        label=f"{employee_count} employees"
    )

# Labels and title
plt.title(
    "Relationship Between Advertising Spend and Revenue",
    fontsize=16
)

plt.xlabel("Advertising Spend (thousand €)")
plt.ylabel("Revenue (thousand €)")

plt.grid(alpha=0.25)

plt.legend(
    title="Industry / Company Size",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()

```
