```python
from graphviz import Digraph



dot=Digraph(format="png")
dot.attr("node",shape="box",style="rounded,filled",fillcolor="lightgrey")

dot.node("A","Data Input (Data Loading)")
dot.node("B","Data Cleaning")
dot.node("C","Data Manipulation & Imputation")
dot.node("D","Data Analysis")
dot.node("E","Data Visualization")
dot.node("F","Predictive Modeling")
dot.node("G","Simulation")
dot.node("H","Reporting")

dot.edge("A","B")
dot.edge("B","C")
dot.edge("C","D")
dot.edge("D","E")
dot.edge("E","B")
dot.edge("E","F")
dot.edge("F","G")
dot.edge("G","H")

dot.render("graph",format="png")
```
